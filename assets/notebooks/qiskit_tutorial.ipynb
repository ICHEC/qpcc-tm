## 2. Creating a Quantum Circuit.

We will use the `QuantumCircuit` class to build circuits. The first argument is the number of qubits in the quantum register, and the second is the number of bits in the classical one.

```python
from qiskit import QuantumCircuit

n = 2 # Number of qubits
qc = QuantumCircuit(n, n) 
```

## 3. Applying Gates to the Circuit

The most common approach to append gates to a circuit is to use named gate methods, like `qc.h(qubit)` (Hadamard gate) , `qc.cx(0,1)` (CNOT gate). Let's build a Bell state.

```python
qc.h(0)
qc.cx(0,1)
```

## 4. Measuring qubits

We can measure our qubits applying `qc.measure(qbit, cbit)` to measure a qubit and store the result into a classical bit. 

```python
for i in range(n):
    qc.measure(i,i)
```

## 5. Drawing circuits

Quantum Circuits can be drawn using : `qc.draw()` method. One can also implement `qc.draw(output='mpl')` to render the circuit as matplotlib figure (*matplotlib* and *pylatexenc* need to be installed)

```python
qc.draw(output='mpl')
```

## 6. Running the circuit in a simulator

### 6.1. Choosing a simulator

The [qiskit-aer package](https://qiskit.github.io/qiskit-aer/index.html) allows us to select and customise a quantum simulator. In this case, we will use a statevector simulator

```python
from qiskit_aer import AerSimulator

simulator = AerSimulator(method="statevector")
```

### 6.2. Transpiling the circuit

Using the `transpile` method allows us to express a circuit into basis gates that the target device or simulator supports. It also applies an optimisation step that reduces gate count and depth, which matters on simulator runtime.

```python
from qiskit import transpile

transpiled_qc = transpile(qc)
```

### 6.3. Running the simulator 

The command `simulator.run()` submits the transpiled circuit to the simualtor backed, and executes the circuit `n_shots` times. Each of the shots indepently samples an outcome according to the circuit's measurements probabilities. 

```python
n_shots = 1024 
job = simulator.run(transpiled_qc, shots=n_shots)
```

### 6.4. Retrieving the results

`job_result()` contains everything about the run: outcomes, metadate, timing, success status... To get the measurement outcomes we can use `result.get_counts()`

```python
result = job.result()
counts = result.get_counts()
print(counts)
```

We can plot the results now: 

```python
import matplotlib.pyplot as plt

counts = result.get_counts()
labels = list(counts.keys())
values = list(counts.values())

plt.figure(figsize=(6, 4))
plt.bar(labels, values)
plt.xlabel("Measurement outcome")
plt.ylabel("Counts")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```
